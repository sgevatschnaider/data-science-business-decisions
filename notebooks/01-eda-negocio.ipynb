## 2. Experimento base

El bloque siguiente construye una referencia mínima y verificable. No representa todavía la recomendación final.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
ventas = pd.DataFrame({
    "canal": rng.choice(["Tienda", "Web", "Marketplace"], 300, p=[0.45, 0.35, 0.20]),
    "ticket": rng.lognormal(mean=3.8, sigma=0.55, size=300),
    "frecuencia": rng.poisson(3, 300),
})
display(ventas.describe(include="all"))
ventas.groupby("canal")["ticket"].agg(["count", "median", "mean"])
```

## 3. Evidencia visual

Una visualización útil permite comparar, muestra unidades y deja visible la incertidumbre o variación relevante.


```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(10, 3.5))
ventas["ticket"].plot.hist(bins=24, ax=axes[0], color="#0f766e", edgecolor="white")
axes[0].axvline(ventas["ticket"].median(), color="#b45309", label="Mediana")
axes[0].set(title="Distribución de ticket", xlabel="Ticket", ylabel="Frecuencia")
axes[0].legend()
ventas.boxplot(column="ticket", by="canal", ax=axes[1], grid=False)
axes[1].set(title="Ticket por canal", xlabel="Canal", ylabel="Ticket")
fig.suptitle("")
plt.tight_layout()
```

## 4. Comparación para decidir

Una cadena minorista analiza ticket, frecuencia, canal y región para decidir dónde concentrar una campaña de fidelización.

La tabla fuerza una comparación entre alternativas, costos o criterios. Adaptala a las unidades del caso.


```python
ventas.groupby('canal').agg(ticket_mediano=('ticket', 'median'), frecuencia_media=('frecuencia', 'mean')).assign(prioridad=lambda x: x['ticket_mediano'] * x['frecuencia_media']).sort_values('prioridad', ascending=False)
```
