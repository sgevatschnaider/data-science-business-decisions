## 2. Experimento base

El bloque siguiente construye una referencia mínima y verificable. No representa todavía la recomendación final.


```python
import random
import numpy as np
import pandas as pd

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

clientes = pd.DataFrame({
    "cliente_id": range(1, 11),
    "antiguedad_meses": np.random.randint(1, 60, 10),
    "compras_90d": np.random.poisson(4, 10),
})
clientes.assign(
    compras_por_mes=lambda df: df["compras_90d"] / 3
).head()
```

## 3. Evidencia visual

Una visualización útil permite comparar, muestra unidades y deja visible la incertidumbre o variación relevante.


```python
import matplotlib.pyplot as plt

ax = clientes.set_index("cliente_id")["compras_90d"].plot.bar(
    figsize=(8, 3), color="#0f766e", title="Compras por cliente"
)
ax.set(xlabel="Cliente", ylabel="Compras en 90 días")
plt.tight_layout()
```
