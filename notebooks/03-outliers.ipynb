## Objetivos

- Detectar extremos con IQR, puntuación z y criterios multivariados.
- Diferenciar outlier, alta palanca e influencia sobre un modelo.
- Aplicar estadísticas robustas y análisis de sensibilidad.
- Justificar conservar, corregir, transformar, limitar o excluir.


```python
import numpy as np
import pandas as pd

ventas = pd.Series([82, 91, 95, 97, 101, 104, 108, 110, 118, 540], name="ticket")
q1, q3 = ventas.quantile([0.25, 0.75])
iqr = q3 - q1
limite_inferior, limite_superior = q1 - 1.5 * iqr, q3 + 1.5 * iqr
diagnostico = pd.DataFrame({
    "ticket": ventas,
    "outlier_iqr": ~ventas.between(limite_inferior, limite_superior),
})
display(diagnostico)
print({"media": ventas.mean(), "mediana": ventas.median()})
```
