## 2. Experimento base

El bloque siguiente construye una referencia mínima y verificable. No representa todavía la recomendación final.


```python
import numpy as np
import pandas as pd

ventas = pd.Series([82, 91, 95, 97, 101, 104, 108, 110, 118, 540], name="ticket")
q1, q3 = ventas.quantile([0.25, 0.75])
iqr = q3 - q1
limite_inferior, limite_superior = q1 - 1.5 * iqr, q3 + 1.5 * iqr
diagnostico = pd.DataFrame({
    "ticket": ventas,
    "outlier_iqr": ~ventas.between(limite_inferior, limite_superior),
})
display(diagnostico)
print({"media": ventas.mean(), "mediana": ventas.median()})
```

## 3. Evidencia visual

Una visualización útil permite comparar, muestra unidades y deja visible la incertidumbre o variación relevante.


```python
import matplotlib.pyplot as plt

colors = diagnostico["outlier_iqr"].map({True: "#be123c", False: "#0f766e"})
fig, ax = plt.subplots(figsize=(8, 3))
ax.scatter(diagnostico.index, diagnostico["ticket"], c=colors, s=70)
ax.axhline(limite_superior, color="#b45309", linestyle="--", label="Límite IQR")
ax.set(xlabel="Observación", ylabel="Ticket", title="Extremos con contexto")
ax.legend()
plt.tight_layout()
```

## 4. Comparación para decidir

Una venta corporativa multiplica por veinte el ticket habitual. Borrarla mejora un promedio, pero elimina el segmento de mayor valor.

La tabla fuerza una comparación entre alternativas, costos o criterios. Adaptala a las unidades del caso.


```python
pd.DataFrame({'escenario': ['todos', 'sin extremo'], 'media': [ventas.mean(), ventas[~diagnostico['outlier_iqr']].mean()], 'mediana': [ventas.median(), ventas[~diagnostico['outlier_iqr']].median()]})
```
